## Dependencies

```python
import os
import torch
from utils import train_UNet_multiclass,BraTsDataset
from model1 import UNet3D
from torch.utils.data import DataLoader

# parameters
device = 'cuda' if torch.cuda.is_available() else 'cpu'
lr = 1e-3
num_epochs=500
patience = 10
batch_size = 4
num_classes = 4
torch.manual_seed(42)

# Paths
train_img_dir = f"../data/brats20/train_data_3channel/train/images/"
train_masks_dir = f"../data/brats20/train_data_3channel/train/masks/"
val_img_dir = f"../data/brats20/train_data_3channel/val/images/"
val_mask_dir = f"../data/brats20/train_data_3channel/val/images/"


os.makedirs('./logs',exist_ok=True)
os.makedirs('./saved_models',exist_ok=True)

model_path_1 = './saved_models/unet3d.pt'
log_path_1 = './logs/train_unet3d.csv'
```

## Load Dataset

```python
train_data = BraTsDataset(img_dir=train_img_dir,
                          img_list=os.listdir(train_img_dir),
                          mask_dir=train_masks_dir,
                          mask_list=os.listdir(train_masks_dir))

val_data = BraTsDataset(img_dir=val_img_dir,
                        img_list=os.listdir(val_img_dir),
                        mask_dir=val_mask_dir,
                        mask_list=os.listdir(val_mask_dir))

train_loader = DataLoader(dataset=train_data,
                          batch_size=batch_size,
                          shuffle=True)
val_loader = DataLoader(dataset=val_data,
                        batch_size=batch_size)
```

```python
for img,mask in train_loader:
    print(img.shape,mask.shape)
    break
```

## Train

### Model 1

```python
model_1 = UNet3D()
train_UNet_multiclass(model=model_1,
                      train_loader=train_loader,
                      valid_loader=val_loader,
                      device=device,
                      num_classes=num_classes,
                      num_epochs=num_epochs,
                      lr=lr,
                      log_path=log_path_1,
                      model_path=model_path_1,
                      patience=patience)
```
